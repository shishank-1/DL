```python
import numpy as np  
import pandas as pd  
import matplotlib.pyplot as plt  

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Input ,Dense , Flatten
from tensorflow.keras.optimizers import SGD, Adam  
from tensorflow.keras.losses import MeanSquaredError , mae
from tensorflow.keras.metrics import Accuracy ,Precision , Recall

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler , MinMaxScaler
```

```python
# Load the dataset (header=None because ecg.csv has no column names)
df = pd.read_csv('ecg.csv', header=None)
df.head()
```

```python
# Data Inspection – df.head(), df.shape, df.info()
# Missing Values Check – df.isnull().sum()
# Duplicate Check – df.duplicated().sum() 
```

```python
# Separate ECG features and labels
# First 140 columns = ECG signals
# Last column = Class label

# `.values` Pandas DataFrame ko NumPy array mein convert karta hai. 
# - Without `.values` → DataFrame format.
# - With `.values` → NumPy array format.

# seperate features and labels
x = df.iloc[:,:-1].values
y = df.iloc[:,-1].values

x
```

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=42)
x_train_normal = x_train[y_train == 1]   # only normal samples for training 1 -- >normal heartbeat

x_train_normal.shape
```

```python
scaler = MinMaxScaler()
x_train_normal = scaler.fit_transform(x_train_normal)
x_test = scaler.fit_transform(x_test)

print(x_train_normal.shape)
print(x_test.shape)
```

```python
# define architecture of Autoencoder 
# Define Encoder (Latent Representation) and Decoder (Reconstruction)

input_dim = x_train.shape[1] #input features 

autoencoder  = Sequential([
    #  # Encoder
    Input(shape=(input_dim,)), # Encoder: Compresses 140 features into 8 dimensions.
    Dense(64,activation='relu'),
    Dense(32,activation='relu'),
    Dense(16,activation='relu'),

    ## Latent Representation
    Dense(8,activation='relu'), 

    ## Decoder
    Dense(16,activation='relu'),
    Dense(32,activation='relu'),
    Dense(64,activation='relu'),

    ## Reconstructs the original 140 features from those 8 dimensions.
    Dense(input_dim,activation='sigmoid') 
    
])

autoencoder.summary()
```

```python
# compile  
autoencoder.compile(
    optimizer= Adam(learning_rate=0.001), 
    loss = MeanSquaredError(),
    metrics=['mae']
)

# train on normal data  
history = autoencoder.fit(
    x_train_normal,x_train_normal , 
    epochs = 10,
    batch_size=512,
    validation_data=(x_test,x_test),
    shuffle=True
)
```

```python
# Plot training and validation loss
plt.plot(history.history['loss'], label='Train Loss')
plt.plot(history.history['val_loss'], label='Validation Loss')
plt.title('Autoencoder Reconstruction Loss')
plt.xlabel('Epochs')
plt.ylabel('Loss (MAE)')
plt.legend()
plt.grid(True)
plt.show()
```

```python
# Step 1: Reconstruct normal training ECG samples
reconstructions = autoencoder.predict(x_train_normal)

# Step 2: Calculate training reconstruction error
# train_loss = #mae(x_train_normal, reconstructions).numpy()
train_loss = np.mean(np.square(x_train_normal-reconstruction),axis=1)

# Step 3: Calculate threshold
threshold = np.mean(train_loss) + np.std(train_loss)

print("Threshold:", threshold)

# Step 4: Reconstruct test ECG samples
pred = autoencoder.predict(x_test)

# Step 5: Calculate reconstruction error for each test sample
error = np.mean((x_test - pred)**2, axis=1)

# Step 6: Predict normal and anomaly
y_pred = (error < threshold).astype(int)
```
